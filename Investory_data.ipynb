```python
# to read dataset
Investory = pd.read_csv("inventory_uncleaned_data.csv")
print("Read Done")
```

```python
# to find null counts
Investory.isnull().sum()
```

```python
# to replace null values
Investory.fillna({"Quantity":"0"},inplace = True)
```

```python
# to replace null values
Investory.fillna({"Price":"0"},inplace = True)
Investory.fillna({"Last Restocked":"0"},inplace = True)
```

```python
# to check null counts
Investory.isnull().sum()
```

```python
# clean the column names
Investory.columns = (Investory.columns.str.strip().str.lower().str.replace(" ", "_"))

print("\nColumn names:")
print(Investory.columns)
```

```python
# Clean text columns
text_columns = [
    "product_name",
    "category",
    "warehouse",
    "location",
    "supplier",
    "status"
]

for column in text_columns:
    Investory[column] = Investory[column].astype("string").str.strip()
```

```python
text_columns
```

```python
# Standardize text
Investory["product_name"] = Investory["product_name"].str.title()

Investory["category"] = Investory["category"].str.upper()

Investory["warehouse"] = Investory["warehouse"].str.title()

Investory["location"] = Investory["location"].str.title()

Investory["supplier"] = Investory["supplier"].str.title()

Investory["status"] = Investory["status"].str.title()

```

```python
Investory
```

```python
# Convert word "two hundred" to 200

Investory["quantity"] = Investory["quantity"].replace(
    {"two hundred": "200"}
)
```

```python
# Convert quantity to numeric
Investory["quantity"] = pd.to_numeric(
    Investory["quantity"],
    errors="coerce"
)
```

```python
# Fill missing quantity with median
Investory["quantity"] = Investory["quantity"].fillna(Investory["quantity"].median()
)
```

```python
# Convert to datatype integer
Investory["quantity"] = Investory["quantity"].astype(int)
```

```python
Investory
```

```python
Investory.describe()
```

```python
# Convert Last Restocked to date
Investory["last_restocked"] = pd.to_datetime(
    Investory["last_restocked"],
    format="%d-%m-%Y",
    errors="coerce"
)
```

```python
Investory
```

```python
# Fill missing dates using the most common date
Investory["last_restocked"] = Investory["last_restocked"].fillna(
    Investory["last_restocked"].mode()[0]
)
```

```python
Investory
```

```python
# Check final missing values
print("\nMissing values after cleaning:")
print(Investory.isnull().sum())
```
